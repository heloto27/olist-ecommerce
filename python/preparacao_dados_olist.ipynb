## 1. Carregamento e exploração inicial de todos os datasets

```python
import glob

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 140)

SRC = "./"  
dfs = {}
for path in glob.glob(os.path.join(SRC, "*.csv")):
    nome = os.path.basename(path).replace(".csv", "")
    dfs[nome] = pd.read_csv(path)

list(dfs.keys())
```

```python
for nome, df in dfs.items():
    print(f"\n{'='*70}\n{nome.upper()}  |  shape: {df.shape}\n{'='*70}")
    print("\n--- head(3) ---")
    display(df.head(3))
    print("\n--- info() ---")
    df.info()
    print("\n--- nulos por coluna ---")
    print(df.isnull().sum())
```

```python
resumo = []
for nome, df in dfs.items():
    resumo.append({
        "tabela": nome,
        "linhas": df.shape[0],
        "colunas": df.shape[1],
        "nulos_total": int(df.isnull().sum().sum()),
        "colunas_com_nulo": int((df.isnull().sum() > 0).sum()),
        "linhas_duplicadas": int(df.duplicated().sum()),
    })

resumo_df = pd.DataFrame(resumo).sort_values("tabela").reset_index(drop=True)
resumo_df
```

## 2. Limpeza de products

```python
df_products = pd.read_csv("olist_products_dataset.csv")

df_products["product_category_name"] = (
    df_products["product_category_name"]
    .fillna("nao_informado")
)

df_products.isnull().sum()
```

```python
# Preenchendo com 0 pois o valor ausente provavelmente indica que o campo
# não foi informado, não que o produto realmente não tem nome/descrição/fotos
colunas_zero = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
]

df_products[colunas_zero] = df_products[colunas_zero].fillna(0)
```

```python
df_products[
    df_products[
        ["product_weight_g",
         "product_length_cm",
         "product_height_cm",
         "product_width_cm"]
    ].isnull().any(axis=1)
]
```

```python
colunas_medidas = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

df_products = df_products.dropna(subset=colunas_medidas)
df_products
```

```python
dfs["olist_products_dataset"] = df_products
```

```python
df_products.isnull().sum()
```

```python
for nome, df in dfs.items():
    print(f"{nome}: {df.isnull().sum().sum()} nulos | {df.duplicated().sum()} duplicatas")
```

## 3. Limpeza de orders

```python
df_orders = pd.read_csv("olist_orders_dataset.csv")
df_orders.isnull().sum()
```

```python
df_orders[df_orders["order_delivered_customer_date"].isnull()]["order_status"].value_counts()
```

**Decisão:** os nulos em `order_delivered_customer_date` correspondem a pedidos
cancelados ou ainda não entregues (confirmado via `order_status`). Foram
